```python
df['hereditary_diseases'] = df['hereditary_diseases'].apply(
    lambda x: 'No' if x == 'NoDisease' else 'Yes'
)
```

```python
dff=pd.get_dummies(df[['sex','hereditary_diseases','city','job_title']],drop_first=True,dtype=int)
```

```python
df_new=pd.concat([df,dff],axis=1)
df_new
```

```python
df_new.drop(['sex','hereditary_diseases','city','job_title'],axis=1,inplace=True)
```

```python
df_new.dtypes
```

```python
x=df_new.drop(['claim'],axis=1)
y=df_new['claim']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.30,random_state=42)
```

```python
feature_columns = x_train.columns.tolist()
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler=MinMaxScaler()
x_train=scaler.fit_transform(x_train)
x_test=scaler.transform(x_test)
```

```python
from sklearn.ensemble import RandomForestRegressor
model=RandomForestRegressor(n_estimators=10)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
y_pred
```

```python
from sklearn.linear_model import LinearRegression
reg_model=LinearRegression()
reg_model.fit(x_train,y_train)
y_pred_reg=reg_model.predict(x_test)
y_pred_reg
```

```python
from sklearn.tree import DecisionTreeRegressor
decision_model=DecisionTreeRegressor()
decision_model.fit(x_train,y_train)
y_pred_decision=decision_model.predict(x_test)
y_pred_decision
```

```python
from sklearn.neighbors import KNeighborsRegressor
knn_model=KNeighborsRegressor()
knn_model.fit(x_train,y_train)
y_pred_knn=knn_model.predict(x_test)
y_pred_knn
```

```python
dfe=pd.DataFrame({'Actual':y_test,'Predicted':y_pred,'Error':y_test-y_pred})
dfe
```

```python
from sklearn.metrics import mean_absolute_percentage_error,r2_score
error=mean_absolute_percentage_error(y_test,y_pred)
error
```

```python
score_decision=r2_score(y_test,y_pred_decision)
score_decision
```

```python
score_knn=r2_score(y_test,y_pred_knn)
score_knn
```
